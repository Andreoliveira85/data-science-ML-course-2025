# Dataset Description
The MPG dataset is technical spec of cars originaly provided from UCI Machine Learning Repository. The data concerns city-cycle fuel consumption in miles per gallon to be analyzed in terms of 3 multivalued discrete and 5 continuous attributes.
Columns Description
- mpg: miles per galon of fuel (continuous variable).
- cylinders: number of engine cylinders (multi-valued discrete variable).
- displacement: (continuous variable)
- horsepower: the power produced by engine to move the car (continuous variable)
- weight: car weight (continuous variable)
- acceleration: the acceleration an engine can get per second (continuous variable)
- model year: car release year from 1970 to 1982(multi-valued discrete variable)
- origin: car manufacturing place (1 -> USA, 2 -> Europe, 3 -> Asia) (multi-valued discrete variable)
- car name: car model name (unique for each instance)

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("auto-mpg.csv")
df
```

```python
df.info()
```

```python
df["car name"].nunique()
```

```python
df.horsepower.unique()
```

```python
df = df.replace("?", np.nan)
```

```python
df.isna().sum()/len(df)
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
df.horsepower = df.horsepower.astype(int)
df.info()
```

```python
## goal: predict mpg

X = df.drop(columns=["mpg", "car name"])
y = df["mpg"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.10, random_state=42)
```

```python
def model_inplace(scaler, model, X_train, X_test, y_train, y_test):
  x_train = scaler.fit_transform(X_train)
  x_test = scaler.transform(X_test)
  model.fit(x_train, y_train)
  print("r2 score on train set", model.score(x_train,y_train))
  print("r2 score on test set", model.score(x_test,y_test))
```

```python
baseline_model = LinearRegression()

## regularisations
ridge = Ridge(alpha = 0.3) ## L2 regulariser
lasso  = Lasso(alpha = 0.1) ## L1 regulariser
```

```python
scaler1 = StandardScaler()
scaler2 = MinMaxScaler()
scaler3 = PolynomialFeatures(degree=2, interaction_only= True) ## check documentation
```

```python
x_train_poly = scaler3.fit_transform(X_train)
x_train_poly
```

```python
x_train_poly.shape ## dimensionality augmentation
```

```python
model_inplace(scaler1, baseline_model, X_train, X_test, y_train, y_test)
```

```python
model_inplace(scaler2, baseline_model, X_train, X_test, y_train, y_test)
```

```python
model_inplace(scaler3, baseline_model, X_train, X_test, y_train, y_test)
## good example of how feature engineering is important in the AI lifecycle
```

```python
model_inplace(scaler1, lasso, X_train, X_test, y_train, y_test)
```

```python
model_inplace(scaler2, lasso, X_train, X_test, y_train, y_test)
```

```python
model_inplace(scaler3, lasso, X_train, X_test, y_train, y_test)
```

```python
## hyper parameter tunning: choosing the best parameters of the model

from sklearn.model_selection import GridSearchCV ## for loop in your parameters

ridge = Lasso()
params = {"alpha": np.arange(0,1000,100)}
#0,10,20,30,40 .... dividing the interval [0,1000] in 100 pieces
grid = GridSearchCV(lasso, param_grid= params, cv =10, verbose=1)
X_train = scaler1.fit_transform(X_train)
X_test = scaler1.transform(X_test)
grid.fit(X_train, y_train)
best_model = grid.best_estimator_
print("coefficients model", best_model.coef_)
print("score on train set", best_model.score(X_train, y_train))
print("score on test set", best_model.score(X_test, y_test))

#feature selection is taking features that are irrelevant to the model

print("columns that were removed by Lasso", X.columns[best_model.coef_==0])
print("columns that were NOT removed by Lasso", X.columns[best_model.coef_ !=0])
print("best alpha", grid.best_params_)
```
