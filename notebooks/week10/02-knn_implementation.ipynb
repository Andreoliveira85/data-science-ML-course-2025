```python
## predictive modelling

## X,y
## train test split
## standardize the data: min max scaler
## fit the model
## evaluate
```

```python
X = df.drop("Class", axis =1)
y = df.Class
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler


scaler = MinMaxScaler()
## X_train = (X_train - min(X_train))/(max(X_train)- min(X_train)) --> fit transform
## X_test = (X_test - min(X_train))/(max(X_train) - min(X_train)) --> transform

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=0)
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# instance a model KNN with 3 neighbours
## trying out blindly 3 neighbours

from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
def modeling(dataset,n):
    X = df.drop("Class", axis =1)
    y = df.Class
    X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=0)
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)
    knn = KNeighborsClassifier(n_neighbors=n)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy =accuracy_score(y_test, y_pred)
    return accuracy
```

```python
modeling(df,3)
```

```python
modeling(df,5)
```

```python
modeling(df,4)
```

```python
modeling(df,9)
```

```python
### choosing the best k model oriented philosophy


accuracy = []

for neighbor in range(1,50):
    knn = KNeighborsClassifier(n_neighbors=neighbor)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy.append(accuracy_score(y_test, y_pred))

plt.figure(figsize=(10,6))
plt.plot(range(1,50), accuracy, color = "blue", linestyle = "dashed",
         marker = "o", markerfacecolor = "red", markersize = 10)

plt.title("accuracy for the best K on the test set")

plt.xlabel("K")
plt.ylabel("accuracy")

print("Maximum accuracy", max(accuracy), "at K", accuracy.index(max(accuracy))+1)
```
