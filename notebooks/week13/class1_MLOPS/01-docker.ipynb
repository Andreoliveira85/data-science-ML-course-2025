## 🔧 Basics of MLOps

**MLOps (Machine Learning Operations)** is the discipline that brings together:
- **Machine Learning**: model training and experimentation.
- **DevOps practices**: automation, testing, monitoring, and deployment.

The goal: **deliver ML models reliably and at scale**.

**Usual components of MLOps:**
- Data pipelines (ETL, feature engineering).
- Experiment tracking (MLflow, Weights & Biases).
- Model versioning and registry.
- Deployment services (APIs, batch jobs, dashboards).
- Monitoring and feedback loops.

Docker plays a key role in the **deployment and reproducibility** parts.

## 🔄 CI/CD Pipeline in MLOps

A **CI/CD (Continuous Integration / Continuous Deployment) pipeline** automates:
1. Integration: testing code & models continuously.
2. Deployment: pushing updates to production reliably.

**Where Docker fits:**
- Ensures consistent environments from development to production.
- Removes the "works on my machine" problem.
- Serves as the base unit for deployment in Kubernetes, cloud, or on-prem.

👉 Think of Docker as the **packaging layer** in your MLOps pipeline.

## 🐳 Why Docker?

- Packages applications + dependencies in **lightweight containers**.
- Containers are **portable** and run the same everywhere.
- Faster and lighter than virtual machines.
- Essential for deploying ML models as APIs, jobs, or dashboards.

> In short: Docker guarantees that what you build locally will run identically in production.

## ⚖️ Containers vs. Virtual Machines

**Virtual Machines (VMs):**
- Run a full operating system.
- Heavy, slow to start.
- Each VM requires dedicated hardware resources.

**Containers:**
- Share the host OS kernel.
- Lightweight and start in seconds.
- Perfect for ML workflows and scaling.

👉 Containers = small, portable, efficient "mini-environments".

## 🧩 Key Docker Concepts

- **Images**: Templates (code, runtime, libraries).
- **Containers**: Running instances of images.
- **Volumes**: Persist data across containers.
- **Ports**: Communication endpoints for networking.
- **Registry (Docker Hub)**: Where images are stored and shared.

👉 Remember: *Images are blueprints, containers are the actual running apps*.

## 📜 Essential Docker Commands

### List running containers
``` docker ps ```

### List all containers (running + stopped)
``` docker ps -a ```

### List images
``` docker images ```

### Remove a container
``` docker rm <container_id> ```

### Remove an image
``` docker rmi <image_id> ```
