## Coding our own linear regression

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression as LinearRegressionSklearn
```

## Linear regression

* (X,y):  y = b + w1X1 + w2 X2 + ...+ wn Xn =  bias + WX

* Gradient descent: calculate the derivative MSE = 1/n \sum(i=1)^n (y_i - pred(x_i))^2

* dMSE/dw = 1/N 2  dot(X,y_i - pred(x_i) )
* dMSE/db  = 1/N sum(y_i - pred(x_i))

* w_t = w_{t-1} + alpha dMSE/dw

```python
class LinearRegression:

    def __init__(self, learning_rate = 0.001, n_iters = 1000):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.weights = None
        self.bias = None
        
    def fit(self, X,y):
        n_samples, n_features = X.shape

        # initialize the parameters
        self.weights = np.zeros(n_features)
        self.bias = 0

        # gradient descent
        for _ in range(self.n_iters):
            y_predicted = self.bias + np.dot(X,self.weights) ## formula of the linear regression

            ## compute the gradients (formulas for the derivative)
            dw = (1/n_samples) * np.dot(X.T, (y_predicted -y))
            db = (1/n_samples) * np.sum(y_predicted -y)

            ## update the parameters (gradient descent formula)
            self.weights -= self.lr * dw 
            self.bias -= self.lr * db
            

    def predict(self,X):
        y_approximated = self.bias + np.dot(X, self.weights)
        return y_approximated
```

```python
## integration test

X = np.random.rand(100,1)
y = 2 + 3 * X.ravel() + np.random.rand(100) ## any Linear regression model needs to learn 2,3

## split the data
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2)
```

```python
## our customized model

model= LinearRegression(learning_rate=0.01, n_iters=10000)
model.fit(X_train, y_train)
y_preds_customized_model = model.predict(X_test)
```

```python
## scikit-learn model

model_sklearn = LinearRegressionSklearn()
model_sklearn.fit(X_train, y_train)
y_preds_sklearn = model_sklearn.predict(X_test)
```

```python
## compare the performances

mse_customized = mean_squared_error(y_test, y_preds_customized_model)
mse_sklearn = mean_squared_error(y_test, y_preds_sklearn)

print("MSE of the first linear model we coded from scratch", mse_customized)
print("MSE of the Sklearn classical model", mse_sklearn)
```
